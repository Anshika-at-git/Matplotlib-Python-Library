```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
scores = np.random.normal(70, 10, 100)

bins = [0, 30, 50, 70, 100]
plt.hist(scores, edgecolor = "black", color = "skyblue", bins = bins)

plt.xlabel("Number of students")
plt.ylabel("Scores")
```

```python
# Multiple Datasets on Histogram

legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(fraud_transactions, color = "red", bins = 10, label = "Fraud", alpha = 0.5, edgecolor = "black")
plt.hist(legit_transactions, color = "green", bins = 10, label = "Legit", alpha = 0.5, edgecolor = "black")

plt.title("Distribution of Transaction Amount - Legit Vs Fraud")
plt.xlabel("Amount (in $)")
plt.ylabel("Frequency")
plt.legend()
```

```python
# Custom Bins

marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

bins = [0, 30, 70, 90, 100]

plt.hist(marks, color = "lavender", bins = bins, edgecolor = "black")
plt.title("Distribution of Marks of Students of a Class")
plt.xlabel("Marks")
plt.ylabel("Frequency")
```

```python
# axvline

marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

bins = [0, 30, 70, 90, 100]

plt.axvline(
    color = "blue",
    x = 33,
    linewidth = 2,
    linestyle = "--",
    label = "Passing Marks (33)"
)

plt.hist(marks, bins = bins, color = "Lavender", edgecolor = "black")

plt.title("Distribution of Marks of Students of a Class")
plt.xlabel("Marks")
plt.ylabel("Frequency")
```
