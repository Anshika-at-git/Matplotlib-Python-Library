```python
import matplotlib.pyplot as plt
```

```python
expenses = ["Salaries", "R&D", "Marketing", "Sales", "Miscellaneous"]
amount = [20, 15, 12, 10, 5] # in $M

plt.style.use("fast")
plt.pie(amount, labels=expenses)
plt.style.available
```

```python
# Customizations on pie chart

colors = ["#C79FEF", "pink", "orange", "#FA8072", "coral"]
explode = [0, 0, 0, 0, 0.25] # Highlights a slice
plt.pie(amount, labels=expenses, autopct="%1.1f%%", colors=colors,
       wedgeprops = {
           "edgecolor" : "black",
           "linestyle" : "-",
           "linewidth" : 1.5,
           "alpha" : 0.9,
           "animated" : True,
           "fill" : True
       }, 
        explode = explode,
        startangle = 360, #Rotates the pie chart by the degree mentioned
        shadow = True
       ) 
```
