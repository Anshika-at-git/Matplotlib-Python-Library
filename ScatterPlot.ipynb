```python
import matplotlib.pyplot as plt
```

```python
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 28, 30, 42, 44, 56, 67, 70, 20]
bp = [110, 115, 118, 130, 135, 115, 137, 140, 144, 112]

plt.scatter(age, bp)
plt.title("Age Vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
```

```python
# Customizations on Scatter Plot

colors = ["green" if x < 130 else "red" for x in bp] # according to colors
plt.scatter(age, bp, color = colors, s = bp, marker = "<", alpha = 0.5) #according to size, marker and transprency
plt.title("BP according to age")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
```

```python
# According to the colormap
plt.scatter(age, bp, s = bp, cmap = "OrRd", c = bp) # c defines the color of colormap
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
plt.colorbar(label="BP") # It is used to show the scale
```

```python
# Annotations

plt.scatter(age, bp, cmap = "twilight", c = bp) # c defines the color of colormap
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i]+1, bp[i]+1))

plt.xlim(min(age)-3, max(age)+10)
plt.ylim(min(bp)-3, max(bp)+10)
```
