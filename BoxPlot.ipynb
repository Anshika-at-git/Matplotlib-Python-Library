```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data)
plt.grid()
```

```python
# Box plot with outliers

data = [7, 8, 5, 6, 9, 4, 10, 12, 15, 20, -3]

plt.boxplot(data)
plt.grid(True)
```

```python
# Multiple DataSets

Group1 = np.random.normal(50, 10, 100)
Group2 = np.random.normal(60, 15, 100)

plt.boxplot([Group1, Group2], tick_labels = ["Group 1", "Group 2"])
plt.grid()
```

```python
# Horizontal Box Plot

data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert = False)
plt.grid()
```

```python
# Showing mean of box plot

data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, showmeans = True)
plt.grid()
```

```python
# Change Whisker Length

data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, whis = 2.0)
plt.grid()
```
