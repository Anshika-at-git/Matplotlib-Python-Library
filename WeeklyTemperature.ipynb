```python
import matplotlib.pyplot as plt
```

```python
# Compare temperature of cities over a week

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 24, 21, 25],
    [18, 19, 17, 20, 21],
    [30, 32, 31, 33, 34],
    [25, 26, 24, 27, 28]
]

fig, ax = plt.subplots(2, 2)
cit = 0
for i in range(2):
    for j in range(2):
        ax[i][j].plot(days, temperatures[cit], marker = "o")
        ax[i][j].set_title(cities[cit])
        ax[i][j].set_xlabel("Days")
        ax[i][j].set_ylabel("Temperature (in C)")
        ax[i][j].grid()
        cit+=1

fig.suptitle("Comparision of City Temperatures")
fig.tight_layout()
```

```python
#Easier way to do it (Using a single loop)

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 24, 21, 25],
    [18, 19, 17, 20, 21],
    [30, 32, 31, 33, 34],
    [25, 26, 24, 27, 28]
]

fig, axes = plt.subplots(2, 2)

axes = axes.flatten()

for i, ax in enumerate(axes):
    ax.plot(days, temperatures[i])
    ax.set_title(cities[i])
    ax.grid()

fig.tight_layout()
```
