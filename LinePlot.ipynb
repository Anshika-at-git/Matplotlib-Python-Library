```python
import matplotlib.pyplot as plt
```

```python
# Basic Example
X = [1, 2, 3, 4]
Y = [5, 6, 7, 8]
plt.plot(X, Y) 
plt.show()
```

```python
# Real Data
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

plt.plot(years, oscar_revenue)
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Revenue of Oscar Winning Movies each year")
```

```python
# Comparing multiple datasets

non_oscar_movies= ["SlumDog Millionaire", "Avatar", "Inception", "Hugo", "LinColn"]
non_oscar_revenue = [378, 2788, 829, 185, 275]

plt.plot(years, oscar_revenue, label = "Oscar Movies")
plt.plot(years, non_oscar_revenue, label = "Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Oscar Vs Non-Oscar Movies Revenue")
plt.legend()
```

```python
# Format Strings
# Syntax -> fmt = [marker][line][color]

plt.plot(years, oscar_revenue,"o--r", label = "Oscar Movies")
plt.plot(years, non_oscar_revenue,">:g", label = "Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Oscar Vs Non-Oscar Movies Revenue")
plt.legend()
```

```python
# Initializing the value individually

plt.plot(years, oscar_revenue,color="red", marker=">", linestyle="-.", linewidth = 2, label = "Oscar Movies")
plt.plot(years, non_oscar_revenue,">:g", label = "Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Oscar Vs Non-Oscar Movies Revenue")
plt.legend()
```

```python
# Styling the line Plot

plt.plot(years, oscar_revenue, label = "Oscar Movies")
plt.plot(years, non_oscar_revenue, label = "Non-Oscar Movies")

plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Oscar Vs Non-Oscar Movies Revenue")

plt.legend()
plt.style.use("ggplot")
plt.grid()
#plt.style.available
```

```python
# To make a Cartoonish graph
with plt.xkcd():
    plt.plot(years, oscar_revenue, label = "Oscar Movies")
    plt.plot(years, non_oscar_revenue, label = "Non-Oscar Movies")
    plt.xlabel("Years")
    plt.ylabel("Revenue (in $M)")
    plt.title("Oscar Vs Non-Oscar Movies Revenue")
    plt.legend()
```

```python
# Comparing movies with revenue
plt.plot(oscar_movies, oscar_revenue)
plt.title("Revenue of Oscar Movies")
plt.xlabel("Movie Names")
plt.ylabel("Revenue ($M)")
plt.savefig("Movie Revenue.png")
plt.tight_layout()
```
